```python
import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
from sklearn.linear_model import LinearRegression,Ridge
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.model_selection import cross_val_score,train_test_split
```

```python
#Download the dataset to this lab environment
def download(url, filename):
    r = requests.get(url)
    if r.status_code == 200:
        with open(filename, "wb") as f:
            f.write(r.content)
        print("File downloaded successfully.")
    else:
        print("File not downloaded. Status Code:", r.status_code)
```

```python
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/medical_insurance_dataset.csv'       
filename = "insurance.csv"
download(url, filename)
```

```python
df = pd.read_csv("insurance.csv")
```

```python
#add the headers to the dataframe 
headers = ["age", "gender", "bmi", "no_of_children", "smoker", "region", "charges"]
df.columns = headers
```

```python
df.head()
```

```python
#Now replace the '?' entries with NaN values
df.replace('?',np.nan,inplace = True)
```

# Task 2 : Data Wrangling

```python
#Use dataframe.info() to identify the columns that have some 'Null' (or NaN) information.
df.info()
```

****Handle missing data:

*For continuous attributes (e.g., age), replace missing values with the mean.*
**For categorical attributes (e.g., smoker), replace missing values with the most frequent value.**
**Update the data types of the respective columns.**
Verify the update using df.info().****

```python
# smoker is a categorical attribute, replace with most frequent entry
is_smoker = df['smoker'].value_counts().idxmax()
df['smoker'].replace(np.nan,is_smoker,inplace = True)

# age is a continuous variable, replace with mean age
mean_age  = df['age'].astype('float').mean()
df['age'].replace(np.nan,mean_age,inplace = True)

```

```python
#update datatype
df[['age','smoker']] = df[['age','smoker']].astype('int')
```

```python
print(df.info())
```

```python
# Update the charges column such that all values are rounded to nearest 2 decimal places. Verify conversion by printing the first 5 values of the updated dataframe.
df[['charges']] = np.round(df[['charges']],2)
df.head()
```
